```python
import random

# ==========================
# DATA
# ==========================

courses = ["CS101", "CS102", "CS103", "CS104"]

faculty = {
    "CS101": "Dr. A",
    "CS102": "Dr. B",
    "CS103": "Dr. A",
    "CS104": "Dr. C"
}

students = {
    "CS101": 40,
    "CS102": 30,
    "CS103": 60,
    "CS104": 25
}

rooms = {
    "RoomA": 50,
    "RoomB": 40,
    "RoomC": 70
}

time_slots = [
    "Mon 9AM",
    "Mon 11AM",
    "Tue 8AM",
    "Tue 11AM",
    "Wed 6PM"
]

POP_SIZE = 6
GENERATIONS = 50
MUTATION_RATE = 0.1

# ==========================
# CREATE RANDOM SCHEDULE
# ==========================

def create_schedule():
    schedule = {}

    for course in courses:
        schedule[course] = (
            random.choice(time_slots),
            random.choice(list(rooms.keys()))
        )

    return schedule


# ==========================
# FITNESS FUNCTION
# ==========================

def fitness(schedule):

    penalty = 0

    course_list = list(schedule.keys())

    for i in range(len(course_list)):
        for j in range(i + 1, len(course_list)):

            c1 = course_list[i]
            c2 = course_list[j]

            time1, room1 = schedule[c1]
            time2, room2 = schedule[c2]

            # Faculty clash
            if faculty[c1] == faculty[c2] and time1 == time2:
                penalty += 20

            # Room clash
            if room1 == room2 and time1 == time2:
                penalty += 20

    # Capacity violation
    for course in courses:

        room = schedule[course][1]

        if students[course] > rooms[room]:
            penalty += 15

    # Optional penalty
    for course in courses:

        slot = schedule[course][0]

        if "8AM" in slot or "6PM" in slot:
            penalty += 5

    return 100 - penalty


# ==========================
# SELECTION
# ==========================

def select(population):

    population.sort(
        key=lambda x: fitness(x),
        reverse=True
    )

    return population[:2]


# ==========================
# CROSSOVER
# ==========================

def crossover(parent1, parent2):

    child = {}

    split = len(courses) // 2

    for i, course in enumerate(courses):

        if i < split:
            child[course] = parent1[course]
        else:
            child[course] = parent2[course]

    return child


# ==========================
# MUTATION
# ==========================

def mutate(schedule):

    if random.random() < MUTATION_RATE:

        course = random.choice(courses)

        schedule[course] = (
            random.choice(time_slots),
            random.choice(list(rooms.keys()))
        )

    return schedule


# ==========================
# INITIAL POPULATION
# ==========================

population = []

for _ in range(POP_SIZE):
    population.append(create_schedule())

print("Initial Population\n")

for i, schedule in enumerate(population):
    print(f"Schedule {i+1}")
    print(schedule)
    print("Fitness =", fitness(schedule))
    print()


# ==========================
# EVOLUTION PROCESS
# ==========================

print("\nEvolution Process\n")

for generation in range(GENERATIONS):

    population.sort(
        key=lambda x: fitness(x),
        reverse=True
    )

    best = population[0]

    print(
        f"Generation {generation+1} "
        f"Fitness = {fitness(best)}"
    )

    parent1, parent2 = select(population)

    new_population = [parent1, parent2]

    while len(new_population) < POP_SIZE:

        child = crossover(parent1, parent2)

        child = mutate(child)

        new_population.append(child)

    population = new_population


# ==========================
# FINAL RESULT
# ==========================

population.sort(
    key=lambda x: fitness(x),
    reverse=True
)

best_schedule = population[0]

print("\n\nFINAL OPTIMIZED WEEKLY SCHEDULE\n")

for course in courses:

    slot, room = best_schedule[course]

    print("Course :", course)
    print("Faculty:", faculty[course])
    print("Time   :", slot)
    print("Room   :", room)
    print()

print("Final Fitness Score =", fitness(best_schedule))
```
