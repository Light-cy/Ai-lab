```python
def count_pattern(pattern,lst):
    count=0
    plen=len(pattern)
    for i in range(len(lst)-plen+1):
        if(tuple(lst[i:i+plen])==tuple(pattern):
            count+=1
    return count
```

```python
#task 1.4.2
class IntegerToRoman:
        # Predefined Roman numeral representations for each digit place
    thousands = ["", "M", "MM", "MMM"]
    hundreds  = ["", "C", "CC", "CCC", "CD", "D", "DC", "DCC", "DCCC", "CM"]
    tens      = ["", "X", "XX", "XXX", "XL", "L", "LX", "LXX", "LXXX", "XC"]
    ones      = ["", "I", "II", "III", "IV", "V", "VI", "VII", "VIII", "IX"]

    def int_to_roman(self, num: int) -> str:
        if(1> num >3999):
            raise ValueError("Input must be between 1 and 3999")
        thousand=self.thousands[(num//1000)]
        hundred=self.hundreds[(num%1000)//100]
        ten=self.tens[(num%100)//10]
        one=self.ones[num%10]
        return thousand+hundred+ten+one
```

```python
def calculator(num1,num2,operation="add",output="integar"):
    if not isinstance(num1,(int,float)) or not isinstance(num2,(int,float)):
        raise TypeError("invalid input")
    result=0
    if(operation=="add"):
        result=num1+num2
    elif(operation=="sub"):
        result=num1-num2
    elif(operation=="div"):
        if(num2==0):
            raise ZeroDivisionError("zero error")
        result=num1/num2
    elif(operation=="mul"):
        result=num1*num2
    else:
        raise ValueError("inavlid operation")
    
    if(output=="integar"):
        return round(result)
    elif(output=="floating"):
        return float(result)
    else:
        raise ValueError("invalid format")
```

```python
class numbers:
    multiplier=10
    def __init__(self,x,y):
        self._x=x
        self._y=y
    
    def add(self):
        return self._x+self._y
    
    @classmethod
    def multiply(self,a):
        return self.multiplier*a
    
    @staticmethod
    def subtract(a,b):
        return a-b
    
    @property
    def value(self):
        return (self._x,self._y)
    
    @value.setter
    def value(self, a):
        self._x,self._y=a
    
    @value.deleter
    def value(self)
        self._x=none
        self._y=none
```
