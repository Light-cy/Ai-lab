```python
def merge(left,right):
    arr1=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            arr1.append(left[i])
            i+=1
        else:
            arr1.append(right[j])
            j+=1

    arr1.extend(left[i:])
    arr1.extend(right[j:])

    return arr1

def mergesort(arr):
    if len(arr)<=1:
        return arr

    mid =len(arr)//2
    left =mergesort(arr[:mid])
    right =mergesort(arr[mid:])
    return merge(left,right)

numbers=[6,3,5,2,4,1]
sortedarray = mergesort(numbers)
print(sortedarray)
```
