```python
def bubblesort(arr):
    if len(arr)<=1:
        return arr
    size = len(arr)

    for i in range(size):
        for j in range(size-1-i):
            if(arr[j]>arr[j+1]):
                temp = arr[j]
                arr[j] = arr[j+1]
                arr[j+1] = temp

    return arr

chars = ['P','Y','T','H','O','N']
print(bubblesort(chars))
```

```python
def selectionsort(arr):
    if len(arr)<=1:
        return arr
    size = len(arr)
    for i in range(size):
        minvalue=i
        for j in range(i+1,size):
            if arr[j]<arr[minvalue]:
                minvalue=j
        arr[i],arr[minvalue]=arr[minvalue],arr[i]
    return arr

chars = ['P','Y','T','H','O','N']
print(selectionsort(chars))
```
